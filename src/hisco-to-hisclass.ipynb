```python
import pandas as pd

# Step 1: Convert CSV file with hisco-hisclass mappings to dictionary
mapping_file_name = 'hisco-to-hisclass.csv'
mappings_df = pd.read_csv(mapping_file_name, dtype={'hisco': str, 'hisclass': int})
hisco_to_hisclass_mapping = mappings_df.set_index('hisco')['hisclass'].to_dict()

# Step 2: Read database with hisco codes to match (e.g. Excel file)
excel_file_name = 'database_hisco.xlsx' #change filename
df = pd.read_excel(excel_file_name, dtype={'hisco': str})

# Step 3: Apply the mappings
df['hisclass'] = df['hisco'].apply(lambda x: hisco_to_hisclass_mapping.get(x, None))

# Step 4: Save the updated data
updated_excel_file_name = 'hisco-hisclass_mappings.xlsx'
df.to_excel(updated_excel_file_name, index=False)

print(f"Hisco-hisclass mappings have been saved as '{updated_excel_file_name}'.")
```
